```python
from math import sqrt, ceil
num= int(input('digite u numero'))
raiz= sqrt(num)
print('a raiz quadrada de {} é {}'.format(num,ceil(raiz)))
```

```python
import math
n= float(input('digite um valor'))
print('o numero {} inteiro fica {}'.format(n,math.trunc(n)))
```

```python
ca= float(input('digite um numero'))
co= float(input('digite um numero'))
h= (ca **2 + co**2)**(1/2)
print('a hipotenusa dos catetos é {}'.format(h))
```

```python
import math
co= float(input("digite um valor "))
ca= float(input('digite um valor '))
h= math.hypot (co, ca)
print(' o valor da hipotenusa é {:.2f}'.format(h))
```

```python
import math
a= float(input('digite um valor: '))
s= math.sin(math.radians(a))
c= math.cos(math.radians(a))
t= math.tan(math.radians(a))
print('o seno de {} é {:.2f}'.format(a,s))
print('o conseno de {} é {:.2f}'.format(a,c))
print('a tangente de {} de é {:.2f}'.format(a,t))
```

```python
import random
a1= str(input('digite um nome: '))
a2= str(input('digite um nome: '))
a3= str(input('digite um nome: '))
a4= str(input('digite um nome: '))
lista= [a1, a2, a3, a4]
escolhido = random.choice(lista)
print('o aluno sorteado foi {}'.format(escolhido))
```

```python
from random import shuffle
a1= str(input('digite um aluno: '))
a2= str(input('digite um aluno: '))
a3= str(input('digite um aluno: '))
a4= str(input('digite um aluno: '))
lista= [a1, a2, a3, a4]
shuffle(lista)
print('a ordem de aluno é {}'.format(lista))
```
