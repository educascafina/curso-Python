Contagem de latas

```python
import math
metros_quadrados= float(input('digite o valor'))
latas = math.ceil(metros_quadrados / 54)
preco= latas * 80
print(f'voce precisa comprar {latas} lata(s), custando R${preco}')
```

Contagem de velocidade

```python
tamanho= float(input('digite o tamanho do arquivo: '))
volocidade= float(input('digite um valor da velocidade: '))
tempo=  (tamanho *8) / volocidade
minuto= tempo // 60
segundo= tempo % 60
print(f'{minuto:.0f} minutos e {segundo:.0f} segundos')
```
