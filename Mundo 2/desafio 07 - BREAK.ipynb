exercicio 01 varios numeros com flag


```python
count= soma= 0
while True:
  number= int(input('digite um numero: '))
  if number == 999:
    break
  count += 1
  soma += number
print(f'a soma dos {count} valores foi {soma}')
```

exercicio 02 tabuada

```python
while True:
  number= int(input('Digite um numero (sera encerrado com um numero negativo): '))
  print('='*70)
  if number < 0:
    break
  for i in range(1,11):
    total= number * i
    print(f'{number} x {i:2} = {total}')
  print('='*70)
print('\033[31mAcabou')
```

exercicio 03 par ou impar

```python
from random import randint
soma= count= 0
while True:
  number= int(input('Choice a number: '))
  computer = randint(0,10)
  soma = computer + number
  choice= ' '
  while choice not in 'IP':
    choice= (str(input('Par ou Impar [I/P]: '))).strip().upper()[0]
  print(f'o computador jogou {computer} e voce {number} o total foi {soma}', end=' ')
  print('DEU PAR'if soma % 2 == 0 else 'DEU IMPAR')
  if soma % 2 == 0:
    if choice in "Pp":
      print('voce ganhou!')
      print('='*50)
    elif choice in 'Ii':
      print('voce perdeu')
      print('='*50)
      break
  else:
    if choice in 'Ii':
      print('voce ganhou!')
      print('='*50)
    elif choice in 'Pp':
      print('Voce perdeu!')
      print('='*50)
      break
  count += 1
print(f'voce venceu {count} vezes consecutivas')
```

exercicio 04 Analise de Dados

```python
cont= contm= contf=  0
while True:
  idade= int(input('Idade: '))
  sexo= ' '
  while sexo not in 'FM':
    sexo= str(input('sexo [F/M]: ')).strip().upper()[0]
    print('='*50)
  if idade >= 18:
    cont += 1
  if sexo == 'M':
    contm += 1
  if sexo == 'F' and idade < 20:
      contf += 1
  pergunta= ' '
  while pergunta not in 'SN':
    pergunta= str(input('Deseja continuar? [S/N]: ')).strip().upper()[0]
    print('='*50)
  if pergunta == 'N':
    break
print(f'Total de pessoas com maior de 18: {cont}')
print(f'total de homens cadastrados: {contm}')
print(f'Tem {contf} mulheres com menos de 20 anos de idade')
```

exercicio 05 estatisticas em produtos

```python
total= maior= menor= cont= 0
barato= ' '
print('{:=^40}'.format('LOJA DO DUDU'))
while True:
  produto= str(input('Nome do Produto: ')).strip().upper()
  preco= float(input('Preço: R$ '))
  cont += 1
  total += preco
  if preco > 1000:
    maior += 1
  if cont == 1 or preco < menor:
    menor = preco
    barato = produto
  res= ' '
  while res not in 'SN':
    res= str(input('Deseja continuar? [S/N]')).strip().upper()[0]
    print('='*40)
  if res == 'N':
    break
print('Compra encerrada')
print(f'O total da sua compra foi: {total:.2f}')
print(f'Houve {maior} produtos que passou de R$ 1.000,00')
print(f'O produto mais barato foi {barato} que custa R$ {menor:.2f}')
```

exercicio 06 caixa eletronico

```python
print('='*50)
print('{: ^40}'.format('BANCO DO DUDU'))
print('='*50)
valor=  int(input('qual é valor que sera sacado?R$ '))
total = valor
ced= 50
totalce= 0
while True:
  if total >= ced:
    total -= ced
    totalce += 1
  else:
    if totalce > 0:
      print(f'o total de {totalce} celulas de R${ced}')
    if ced == 50:
      ced = 20
    if ced == 20:
      ced = 10
    if ced == 10:
      ced = 1
    totalce += 0
    if total == 0:
      break
print('='*50)
print('acabou volte sempre')


```
