exercicio 02 jogo da advinhaçao

```python
from random import randint
from time import sleep
computer= randint(0,10)
soma= 1
print('Olá!, sou seu computador e vou chutar um numero e vc tentara adivinhar...')
sleep(2)
player= int(input('Qual é seu palpite: '))
while player != computer:
  if player > computer:
    player= int(input('Menos... tente novamente: '))
  if player < computer:
    player= int(input('Mais... tente novamente: '))
  soma += 1
print('Voce acertou o com {} palpites. PARABENS!'.format(soma))
```

exercicio 03 menu de opcao

```python
from time import sleep
n1= int(input('Primeiro numero: '))
n2= int(input('Segundo numero: '))
opcao= 0
while opcao != 5:
    print('='*50)
    print('''
         [1] Somar
         [2] Multiplicar
         [3] Maior
         [4] Novos numeros
         [5] Sair do programa''')
    opcao= int(input('>>>>>>>>> Qual sua opçao? '))
    if opcao == 1:
      print('A soma entre {} + {} é {}'.format(n1, n2, n1+n2))
    elif opcao == 2:
      print('A multiplicao entre {} x {} é {}'.format(n1, n2, n1*n2))
    elif opcao == 3:
      if n1 > n2:
        maior= n1
      else:
        maior = n2
      print('Entre o {} e o {} o maior é {}'.format(n1, n2, maior))
    elif opcao == 4:
      print('Digite os numeros novamente:')
      n1= int(input('Primeiro numero: '))
      n2= int(input('Segundo numero: '))
    elif opcao == 5:
      print('Finalizando...')
      sleep(2)
    else:
      print('Numero invalido! tente novamente')
print('='*50)
print('Programa encerrado!')
```

exercio 05 progressao artimetica

```python
print('='*30)
print('Progressão Aritmetica')
print('='*30)
p= int(input('Primeiro termo: '))
r= int(input('Razão: '))
d= p
c= 1
while c <= 10 :
  print(' {} -->'.format(d), end=' ')
  d += r
  c += 1
print('Fim')
```

exercicio 06 super P.A

```python
print('='*30)
print('Progressão Aritmetica')
print('='*30)
primeiro= int(input('Primeiro Termo: '))
razao= int(input('Razão: '))
variavel= primeiro
count= 1
total= 0
second = 10
while second != 0:
  total = total + second
  while count <= total :
    print(' {} -->'.format(variavel), end=' ')
    variavel += razao
    count += 1
  print('Pause')
  second= int(input('Quantos termos voce quer mais? '))
print('Progressao finalizada com o total de {} termos'.format(total))
```

exercicio 07 Sequência de Fibonacci

```python
termo= int(input('Quantos termos voce quer mostrar? '))
first= 0
second= 1
print('{} -> {} ->'.format(first, second), end=' ')
count= 3
while count <= termo:
  third= first + second
  print('{} ->'.format(third), end=' ')
  first = second
  second = third
  count += 1
print('Fim')
```
