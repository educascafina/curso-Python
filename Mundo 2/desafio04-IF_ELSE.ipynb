Exercicio 01
PRESTACAO DE CASA

```python
valor = float(input('digite o valor do imovel:R$ '))
salario = float(input('digite o salario do empregador:R$ '))
anos = int(input('digite em quantos anos voce ira pagar: '))
prestacao = valor / (anos * 12)
minimo = salario * 30 / 100
print('para pagar uma casa de {:.2f} em {} anos'.format(valor, anos), end="")
print(' a prestacao sera R${:.2f}'.format(prestacao))
if prestacao <= minimo:
  print('valor aprovado')
else:
  print("valor negado!")
```

Exercicio 02
CONVERTER EM BI/OCT/HEX

```python
n= int(input('digite um valor: '))
print('''escolha como vai querer sua conversao:
[1]coverter para binario
[2]converter para octal
[3]converter para hexadecimal''')
opcao = int(input('sua opcao: '))
if opcao == 1:
  print('{} convertido para binario fica {}'.format(n, bin(n)[2:]))
elif opcao == 2:
  print('{} convertido para octal fica {}'.format(n, oct(n)[2:]))
elif opcao == 3:
  print('{} convertido para hexadecimal fica {}'.format(n, hex(n)[2:]))
else:
  print('opcao invalida, tente novamente!')
```

EXERCICIO 04  ALISTAMENTO

```python
from datetime import date
a= date.today().year
ano= int(input('digite o ano do seu nascimento: '))
idade= a - ano
print('quem nasceu no ano de {} tem {} anos em {}'.format(ano, idade, a))
if idade == 18:
  print('voce deve se alistar')
elif idade < 18:
  falta= 18 - idade
  print('nao é a hora ainda, faltam {} anos'.format(falta))
  qnd= a + falta
  print('seu alistamento vai ser em {}'.format(qnd))
else:
  falta= idade - 18
  print('ja passou do prazo, o seu alistamento foi {} anos tras'.format(falta))
  qnd= a - falta
  print('voce deveria ter se alistado em {}'.format(qnd))
```

exercicio 06 triangulos

```python
print('-='*20)
print('analisador de triangulo')
print('-='*20)
r1= float(input('digite o valor: '))
r2= float(input('digite o valor: '))
r3= float(input('digite um valor: '))
if r1 < r2 + r3 and r2 < r3 + r1 and r3 < r1 + r2:
  print('esse segmento forma um triangulo ', end='')
  if r1 == r2 == r3:
    print('equiletero')
  elif r1 != r2 != r3 != r1:
    print('escaleno')
  else:
    print('isosceles')
else:
  print('esse segmento nao forma um triangulo')
```

exercicio 07 IMC

```python
p= float(input('digite seu peso: (kg) '))
a= float(input(' digite sua altura: (m) '))
imc= p / (a ** 2)
print('seu IMC é {:.2f} '.format(imc), end='')
if imc < 18.5:
  print(' e esta abaixo do peso')
elif 18.5 <= imc < 25:
  print(' e esta no peso ideal')
elif 25 <= imc < 30:
  print(' e esta no sobrepeso')
elif 30 <= imc < 40:
  print(' e esta na obsidade')
else:
  print(' e esta na obsidade morbida')
```
