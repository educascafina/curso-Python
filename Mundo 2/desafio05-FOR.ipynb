exercicio 07 numeros primos

```python
n= int(input(' Digite um numero: '))
tot= 0
for c in range (1, n+1):
  if n % c == 0:
    print('\033[33m', end=' ')
    tot +=1
  else:
    print('\033[31m', end=' ')
  print('{}'.format(c), end=' ')
print('\n\033[0mo numero {} foi divisivel {} vezes'.format(n, tot))
if tot==2:
  print('é por isso que é PRIMO')
else:
  print('é por isso que ele \033[31mNÃO É PRIMO')
```

exercicio 08 palíndromo

```python
f= str(input('Digite uma frase: ')).strip().upper()
palavras= f.split()
juntar= ''.join(palavras)
inverso = ''  # da pra fazer sem o for (inverso[::-1]) e pronto
for l in range(len(juntar) -1, -1, -1):
  inverso += juntar[l]
print(' o inverso de {} é {}'.format(juntar,inverso))
if inverso == juntar:
  print(' é um palíndromo')
else:
  print(' \033[31mNao e um palíndromo')
```

exercicio 09 maioridade

```python
from datetime import date
m = date.today().year
totmaior= 0
totmenor= 0
for i in range (1, 8):
  ano= int(input('digite o {}° ano de nascimento: '.format(i)))
  id= m - ano
  if id >= 21:
    totmaior += 1
  else:
    totmenor += 1
print('O total de pessoa maior de idade é {}'.format(totmaior))
print('O total de pessoa menor de idade é {}'.format(totmenor))
```

exercicio 10 maior e menor peso

```python
maior= 0
menor=0
for p in range(1, 6):
  peso= float(input('digite o {}° peso: '.format(p)))
  if  p == 1:
    maior = peso
    menor = peso
  else:
    if peso > maior:
      maior = peso
    if peso < menor:
      menor = peso
print('o maior peso lido é de {}KG'.format(maior))
print('o menor peso lido é de {}KG'.format(menor))
```
